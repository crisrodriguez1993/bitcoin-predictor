```python
# Conexión de Google Drive y definición de ruta del proyecto
from google.colab import drive
drive.mount('/content/drive')

PROJECT_ROOT = '/content/drive/MyDrive/01. USFQ/01. Maestría IA/12. Trabajo de Titulación/05. V2'
print('PROJECT_ROOT =', PROJECT_ROOT)
```

```python
# Clonar/sincronizar rama específica (dev) en Drive
from pathlib import Path

REPO_URL = 'https://github.com/crisrodriguez1993/bitcoin-predictor.git'
REPO_BRANCH = 'dev'  # Cambia aquí si luego deseas otra rama
project_path = Path(PROJECT_ROOT)

if not project_path.exists():
    project_path.mkdir(parents=True, exist_ok=True)

if not (project_path / '.git').exists():
    %cd "{project_path.parent}"
    !git clone --branch "{REPO_BRANCH}" --single-branch "{REPO_URL}" "{project_path.name}"
    print(f'Repositorio clonado en {project_path} (rama: {REPO_BRANCH})')
else:
    print('Repositorio ya existe. Sincronizando rama...')
    !git -C "{project_path}" fetch --all --prune
    !git -C "{project_path}" checkout "{REPO_BRANCH}"
    !git -C "{project_path}" pull origin "{REPO_BRANCH}"
    print(f'Repositorio sincronizado en {project_path} (rama: {REPO_BRANCH})')
```
